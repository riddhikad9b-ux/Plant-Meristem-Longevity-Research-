````python
# ============================================================
# STEP 2 - ORGANIZE miRNA DATA BY TREE AGE
# ============================================================

young_samples = [
"VC20A.readcount",
"VC20B.readcount"
]

middle_samples = [
"VC200A.readcount",
"VC200B.readcount"
]

old_samples = [
"VC600A.readcount",
"VC600B.readcount"
]

# Check required columns WITHOUT using if/else
required_columns = (
["miRNA_name"]
+ young_samples
+ middle_samples
+ old_samples
)

missing_columns = [
col for col in required_columns
if col not in df.columns
]

print("Missing columns:", missing_columns)

assert not missing_columns, "Some required columns are missing."

print("All expected columns found.")

# Calculate age-group means
df["Young_mean"] = df[young_samples].mean(axis=1)

df["Middle_mean"] = df[middle_samples].mean(axis=1)

df["Old_mean"] = df[old_samples].mean(axis=1)

# Calculate fold changes
df["Old_vs_Young_FC"] = (
(df["Old_mean"] + 1) /
(df["Young_mean"] + 1)
)

df["Middle_vs_Young_FC"] = (
(df["Middle_mean"] + 1) /
(df["Young_mean"] + 1)
)

# Display first 10 rows
display(
df[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"Middle_vs_Young_FC"
]
].head(10)
)
````

````python
# ============================================================
# STEP 4 - COMPLETE miR156 ANALYSIS TABLE
# ============================================================

miR156_analysis = miR156_df[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"Middle_vs_Young_FC"
]
].copy()

# Add absolute change from Young to Old
miR156_analysis["Old_minus_Young"] = (
miR156_analysis["Old_mean"]
- miR156_analysis["Young_mean"]
)

# Sort by Old vs Young fold-change
miR156_analysis = miR156_analysis.sort_values(
"Old_vs_Young_FC",
ascending=False
)

print("Complete miR156 analysis:")
display(miR156_analysis)
````

````python
# ==========================================
# STEP 8 — GROUP STRONG miRNAs BY FAMILY
# ==========================================

family_df = strong_miRNAs.copy()

# Extract miRNA family directly
family_df["Family"] = (
family_df["miRNA_name"]
.astype(str)
.str.lower()
.str.extract(r"(mir\d+)", expand=False)
)

# Count miRNAs in each family
family_counts = (
family_df["Family"]
.value_counts()
.reset_index()
)

family_counts.columns = ["Family", "Count"]

print("miRNA family counts:")
display(family_counts)

print("\nStrong age-related miRNAs with families:")

display(
family_df[
[
"miRNA_name",
"Family",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"Middle_vs_Young_FC"
]
]
)
````

````python
# ==========================================
# STEP 1 — Prepare miRNA expression data
# ==========================================

import pandas as pd
import numpy as np

# Use the dataframe that actually exists
age_miRNA = mirna_df.copy()

print("Shape:", age_miRNA.shape)
print("\nColumns:")
print(list(age_miRNA.columns))

print("\nFirst 5 rows:")
display(age_miRNA.head())
````

````python
# ==========================================
# STEP 2 — Calculate mean expression
# ==========================================

age_miRNA["Young_mean"] = age_miRNA[
["VC20A.readcount", "VC20B.readcount"]
].mean(axis=1)

age_miRNA["Middle_mean"] = age_miRNA[
["VC200A.readcount", "VC200B.readcount"]
].mean(axis=1)

age_miRNA["Old_mean"] = age_miRNA[
["VC600A.readcount", "VC600B.readcount"]
].mean(axis=1)

print("Calculated age-group means successfully.")

display(
age_miRNA[
["miRNA_name", "Young_mean", "Middle_mean", "Old_mean"]
].head(10)
)
````

````python
# ==========================================
# STEP 3 — Calculate age-related changes
# ==========================================

age_miRNA["log2_Old_vs_Young"] = np.log2(
(age_miRNA["Old_mean"] + 1) /
(age_miRNA["Young_mean"] + 1)
)

age_miRNA["log2_Old_vs_Middle"] = np.log2(
(age_miRNA["Old_mean"] + 1) /
(age_miRNA["Middle_mean"] + 1)
)

# Sort by strongest Old vs Young change
age_miRNA_sorted = age_miRNA.sort_values(
"log2_Old_vs_Young",
ascending=False
)

print("Age-related analysis completed.")

display(
age_miRNA_sorted[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"log2_Old_vs_Young",
"log2_Old_vs_Middle"
]
].head(20)
)
````

````python
# ==========================================
# STEP 4 — Filter stronger age-associated miRNAs
# ==========================================

# Calculate average expression across all age groups
age_miRNA["Overall_mean"] = age_miRNA[
["Young_mean", "Middle_mean", "Old_mean"]
].mean(axis=1)

# Calculate fold change
age_miRNA["Old_vs_Young_FC"] = (
(age_miRNA["Old_mean"] + 1) /
(age_miRNA["Young_mean"] + 1)
)

# Keep miRNAs with reasonable expression
# and at least 2-fold change in either direction
candidate_miRNAs = age_miRNA[
(age_miRNA["Overall_mean"] >= 10) &
(abs(age_miRNA["log2_Old_vs_Young"]) >= 1)
].copy()

# Sort by magnitude of age-related change
candidate_miRNAs["abs_log2FC"] = abs(
candidate_miRNAs["log2_Old_vs_Young"]
)

candidate_miRNAs = candidate_miRNAs.sort_values(
"abs_log2FC",
ascending=False
)

print("Number of candidate miRNAs:", len(candidate_miRNAs))

display(
candidate_miRNAs[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"log2_Old_vs_Young"
]
].head(30)
)
````

````python
# ==========================================
# STEP 5 — Separate increased and decreased
# ==========================================

# miRNAs increased in Old compared with Young
increased_miRNAs = candidate_miRNAs[
candidate_miRNAs["log2_Old_vs_Young"] >= 1
].sort_values(
"log2_Old_vs_Young",
ascending=False
)

# miRNAs decreased in Old compared with Young
decreased_miRNAs = candidate_miRNAs[
candidate_miRNAs["log2_Old_vs_Young"] <= -1
].sort_values(
"log2_Old_vs_Young",
ascending=True
)

print("Increased in Old:", len(increased_miRNAs))
print("Decreased in Old:", len(decreased_miRNAs))

print("\n===== TOP INCREASED =====")

display(
increased_miRNAs[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"log2_Old_vs_Young"
]
].head(20)
)

print("\n===== TOP DECREASED =====")

display(
decreased_miRNAs[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"log2_Old_vs_Young"
]
].head(20)
)
````

````python
# ============================================================
# RESEARCH LAB v3 - STAGE 2: C-ALPHA DISTANCE MATRIX & ANALYSIS
# ============================================================

import numpy as np
from Bio.PDB import PDBParser

print("=" * 70)
print("INITIALIZING RESEARCH LAB v3 - STAGE 2: DISTANCE MATRIX")
print("=" * 70)

# Load the previously downloaded structure file
pdb_id = "1LCD"
structure_file = f"pdb{pdb_id.lower()}.ent" # Standard naming convention from PDBList

parser = PDBParser(QUIET=True)
structure = parser.get_structure(pdb_id, structure_file)

# Extract all C-alpha (CA) atoms from the first model
model = structure[0]
ca_atoms = []
residue_labels = []

for chain in model:
    for residue in chain:
        # Check if C-alpha atom exists in the residue
        if "CA" in residue:
            ca_atoms.append(residue["CA"])
            # Label with Chain ID and Residue Number
            residue_labels.append(f"{chain.id}:{residue.resname}{id(residue) % 1000}")

print(f"Extracted {len(ca_atoms)} C-alpha atoms for distance calculations.")

# Compute pairwise Euclidean distance matrix between C-alpha atoms
num_atoms = len(ca_atoms)
distance_matrix = np.zeros((num_atoms, num_atoms))

for i in range(num_atoms):
    for j in range(num_atoms):
        distance_matrix[i, j] = ca_atoms[i] - ca_atoms[j]

print("-" * 50)
print(f"DISTANCE MATRIX COMPUTED SUCCESSFULLY: shape {distance_matrix.shape}")
print(f"  - Minimum non-zero distance: {np.min(distance_matrix[distance_matrix > 0]):.2f} Å")
print(f"  - Maximum distance: {np.max(distance_matrix):.2f} Å")
print(f"  - Mean pairwise distance: {np.mean(distance_matrix):.2f} Å")
print("-" * 50)

print("=" * 70)
print("RESEARCH LAB v3 STAGE 2 COMPLETED SUCCESSFULLY")
print("=" * 70)
````

````python
# Install required libraries if missing in this runtime session
!pip install -q biopython

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from Bio.PDB import PDBParser, PDBList

# 1. Self-contained PDB loading and parsing
pdb_id = "1LCD"
pdbl = PDBList()
pdbl.retrieve_pdb_file(pdb_id, pdir='.', file_format='pdb')
structure_file = f"pdb{pdb_id.lower()}.ent"

parser = PDBParser(QUIET=True)
structure = parser.get_structure(pdb_id, structure_file)

# 2. Extract C-alpha coordinates and compute distance matrix
ca_atoms = []
for model in structure:
    for chain in model:
        for residue in chain:
            if "CA" in residue:
                ca_atoms.append(residue["CA"])

coords = np.array([atom.get_coord() for atom in ca_atoms])
distance_matrix = np.sqrt(np.sum((coords[:, np.newaxis, :] - coords[np.newaxis, :, :]) ** 2, axis=-1))

# 3. Research Lab v4: Binary Contact Map Analysis
contact_threshold = 8.0
contact_map = np.where(distance_matrix <= contact_threshold, 1, 0)
np.fill_diagonal(contact_map, 0)

# 4. Plotting the binary contact map
plt.figure(figsize=(8, 6))
sns.heatmap(contact_map, cmap="Blues", cbar=True, square=True, linewidths=0, linecolor='gray')
plt.title(f"Research Lab v4: C-Alpha Residue Contact Map ({pdb_id})", fontsize=12, pad=12)
plt.xlabel("Residue Index", fontsize=10)
plt.ylabel("Residue Index", fontsize=10)
plt.show()

# 5. Quantitative Summary
total_contacts = np.sum(contact_map) // 2
print("-" * 50)
print(f"📊 Structural Analysis Summary:")
print(f"   - Contact Distance Threshold: {contact_threshold} Å")
print(f"   - Total Unique Residue-Residue Contacts: {total_contacts}")
print("-" * 50)
````

````python
import numpy as np

# Use the upper triangle of the contact map to avoid duplicate pairs and self-loops
upper_triangle = np.triu(contact_map, k=1)
res_i, res_j = np.where(upper_triangle == 1)

print(f"📊 Interacting Residue Network List ({pdb_id}):")
print(f"   - Total Contact Pairs Found: {len(res_i)}")
print("-" * 50)
print("   Sample Interacting Pairs (First 15):")

# Print the first 15 pairs for quick inspection
for idx, (i, j) in enumerate(zip(res_i[:15], res_j[:15]), start=1):
    # Optional: Calculate actual distance for these specific pairs
    actual_dist = distance_matrix[i, j]
    print(f"   {idx:2d}. Residue {i+1} <---> Residue {j+1}  ({actual_dist:.2f} Å)")

print("-" * 50)
````

````python
# Install networkx if not already present in the runtime session
!pip install -q networkx

import networkx as nx
import numpy as np

# Build a NetworkX graph directly from the binary contact map
G = nx.from_numpy_array(contact_map)

# Calculate degree centrality (relative measure of connectivity for each residue)
degree_centrality = nx.degree_centrality(G)

# Identify top hub residues sorted by their connection density
top_hubs = sorted(degree_centrality.items(), key=lambda x: x[1], reverse=True)[:10]

print("-" * 50)
print(f"🧬 Research Lab v5: Residue Contact Network Analysis ({pdb_id})")
print(f"   - Total Graph Nodes (Residues): {G.number_of_nodes()}")
print(f"   - Total Graph Edges (Contacts): {G.number_of_edges()}")
print("-" * 50)
print("   Top 10 Structural Hub Residues (by Degree Centrality):")
for rank, (residue_idx, centrality) in enumerate(top_hubs, start=1):
    contact_count = G.degree(residue_idx)
    print(f"   {rank:2d}. Residue {residue_idx+1} | Centrality: {centrality:.3f} | Contacts: {contact_count}")
print("-" * 50)
````

````python
import matplotlib.pyplot as plt
import networkx as nx

# Set up the figure for network graph plotting
plt.figure(figsize=(10, 8))

# Use spring layout to position nodes based on physical connectivity
pos = nx.spring_layout(G, seed=42)

# Scale node sizes based on their degree centrality (larger nodes = major hubs)
node_sizes = [v * 1500 for v in degree_centrality.values()]

# Draw the network graph components (fixed edgecolor -> edgecolors)
nx.draw_networkx_nodes(G, pos, node_size=node_sizes, node_color='skyblue', alpha=0.8, edgecolors='black')
nx.draw_networkx_edges(G, pos, alpha=0.2, edge_color='gray')

# Highlight top 3 hubs with a distinct color (e.g., crimson)
top_3_nodes = [node for node, _ in top_hubs[:3]]
nx.draw_networkx_nodes(G, pos, nodelist=top_3_nodes, node_size=1200, node_color='crimson', alpha=0.9, edgecolors='black')

# Graph layout configurations
plt.title(f"Research Lab v5: Residue Contact Network Topology ({pdb_id})", fontsize=13, pad=15)
plt.axis('off')
plt.tight_layout()
plt.show()

print("-" * 50)
print("🌐 Contact Network Topology Graph Generated Successfully!")
print("   (Crimson nodes highlight top structural stability hubs)")
print("-" * 50)
````

````python
import matplotlib.pyplot as plt
import numpy as np

# Extract B-factors
b_factors = [atom.get_bfactor() for atom in ca_atoms]

# Fallback: if PDB file has 0.0 B-factors, use coordinate deviation from mean as a thermal/positional proxy
if all(bf == 0.0 for bf in b_factors):
    coords = np.array([atom.get_coord() for atom in ca_atoms])
    mean_coord = np.mean(coords, axis=0)
    b_factors = np.sqrt(np.sum((coords - mean_coord) ** 2, axis=1))
    metric_label = "Positional Deviation Proxy (Å)"
else:
    metric_label = "B-Factor (Å²)"

residue_indices = list(range(1, len(b_factors) + 1))
mean_val = np.mean(b_factors)
max_val = np.max(b_factors)
min_val = np.min(b_factors)

# Plotting the flexibility profile
plt.figure(figsize=(10, 4))
plt.plot(residue_indices, b_factors, color='purple', linewidth=1.5, label=metric_label)
plt.axhline(y=mean_val, color='black', linestyle='--', alpha=0.7, label=f'Mean ({mean_val:.2f})')
plt.title(f"Research Lab v6: Structural Flexibility Profile ({pdb_id})", fontsize=12, pad=12)
plt.xlabel("Residue Index", fontsize=10)
plt.ylabel(metric_label, fontsize=10)
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()

print("-" * 50)
print(f"🌡️ Structural Flexibility Analysis Summary ({pdb_id}):")
print(f"   - Metric Used: {metric_label}")
print(f"   - Mean Value:  {mean_val:.2f}")
print(f"   - Minimum:     {min_val:.2f}")
print(f"   - Maximum:     {max_val:.2f}")
print("-" * 50)
````

````python
# 5. Perform Superimposition Demonstration
superimposer = Superimposer()

# For a true comparison, you would load a variant PDB file.
# Here we map the reference against itself (RMSD should be 0.00 Å)
# as a validation of the alignment pipeline.
superimposer.set_atoms(ca_ref, ca_ref)
superimposer.apply(structure_ref.get_atoms())

print(f"📊 Structural Alignment Complete!")
print(f" Root Mean Square Deviation (RMSD): {superimposer.rms:.4f} Å")
print(f" Rotation Matrix:\n {superimposer.rotran[0]}")
print(f" Translation Vector: {superimposer.rotran[1]}")

````

````python
import pandas as pd

def add_research_paper(paper_id, doi_pmid, year, species, tissue, gene_pathway,
                       research_question, experimental_method, major_result,
                       limitation, research_gap, confidence_label):
    df = pd.read_csv("research_matrix.csv")
    new_row = {
        "Paper_ID": paper_id,
        "DOI_PMID": doi_pmid,
        "Year": year,
        "Species": species,
        "Tissue": tissue,
        "Gene_Pathway": gene_pathway,
        "Research_Question": research_question,
        "Experimental_Method": experimental_method,
        "Major_Result": major_result,
        "Limitation": limitation,
        "Research_Gap": research_gap,
        "Confidence_Label": confidence_label
    }
    df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
    df.to_csv("research_matrix.csv", index=False)
    print(f"Successfully added {paper_id} to the research matrix!")

add_research_paper(
    paper_id="Paper_02",
    doi_pmid="10.1038/s41586-xxxx",
    year="2020",
    species="Arabidopsis thaliana",
    tissue="Meristematic tissue",
    gene_pathway="Heterotrimeric G-protein signaling",
    research_question="How do G-protein subunits modulate abiotic stress memory in meristems?",
    experimental_method="Stress-priming assays and transcriptional tracking",
    major_result="G-protein mutants show altered tolerance memory upon repeated stress exposure.",
    limitation="Tested under acute short-term stress rather than multi-decade aging.",
    research_gap="Whether stress memory pathways contribute to long-term cellular resilience in centenarian trees.",
    confidence_label="BLUE (supported association)"
)
````
