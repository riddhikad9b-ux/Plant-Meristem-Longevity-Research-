```python
# ============================================================
# STEP 2 - ORGANIZE miRNA DATA BY TREE AGE
# ============================================================

young_samples = [
"VC20A.readcount",
"VC20B.readcount"
]

middle_samples = [
"VC200A.readcount",
"VC200B.readcount"
]

old_samples = [
"VC600A.readcount",
"VC600B.readcount"
]

# Check required columns WITHOUT using if/else
required_columns = (
["miRNA_name"]
+ young_samples
+ middle_samples
+ old_samples
)

missing_columns = [
col for col in required_columns
if col not in df.columns
]

print("Missing columns:", missing_columns)

assert not missing_columns, "Some required columns are missing."

print("All expected columns found.")

# Calculate age-group means
df["Young_mean"] = df[young_samples].mean(axis=1)

df["Middle_mean"] = df[middle_samples].mean(axis=1)

df["Old_mean"] = df[old_samples].mean(axis=1)

# Calculate fold changes
df["Old_vs_Young_FC"] = (
(df["Old_mean"] + 1) /
(df["Young_mean"] + 1)
)

df["Middle_vs_Young_FC"] = (
(df["Middle_mean"] + 1) /
(df["Young_mean"] + 1)
)

# Display first 10 rows
display(
df[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"Middle_vs_Young_FC"
]
].head(10)
)
```

```python
# ==========================================
# STEP 8 — GROUP STRONG miRNAs BY FAMILY
# ==========================================

family_df = strong_miRNAs.copy()

# Extract miRNA family directly
family_df["Family"] = (
family_df["miRNA_name"]
.astype(str)
.str.lower()
.str.extract(r"(mir\d+)", expand=False)
)

# Count miRNAs in each family
family_counts = (
family_df["Family"]
.value_counts()
.reset_index()
)

family_counts.columns = ["Family", "Count"]

print("miRNA family counts:")
display(family_counts)

print("\nStrong age-related miRNAs with families:")

display(
family_df[
[
"miRNA_name",
"Family",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"Middle_vs_Young_FC"
]
]
)
```

```python
# ==========================================
# STEP 1 — Prepare miRNA expression data
# ==========================================

import pandas as pd
import numpy as np

# Use the dataframe that actually exists
age_miRNA = mirna_df.copy()

print("Shape:", age_miRNA.shape)
print("\nColumns:")
print(list(age_miRNA.columns))

print("\nFirst 5 rows:")
display(age_miRNA.head())
```

```python
# ==========================================
# STEP 2 — Calculate mean expression
# ==========================================

age_miRNA["Young_mean"] = age_miRNA[
["VC20A.readcount", "VC20B.readcount"]
].mean(axis=1)

age_miRNA["Middle_mean"] = age_miRNA[
["VC200A.readcount", "VC200B.readcount"]
].mean(axis=1)

age_miRNA["Old_mean"] = age_miRNA[
["VC600A.readcount", "VC600B.readcount"]
].mean(axis=1)

print("Calculated age-group means successfully.")

display(
age_miRNA[
["miRNA_name", "Young_mean", "Middle_mean", "Old_mean"]
].head(10)
)
```

```python
# ==========================================
# STEP 3 — Calculate age-related changes
# ==========================================

age_miRNA["log2_Old_vs_Young"] = np.log2(
(age_miRNA["Old_mean"] + 1) /
(age_miRNA["Young_mean"] + 1)
)

age_miRNA["log2_Old_vs_Middle"] = np.log2(
(age_miRNA["Old_mean"] + 1) /
(age_miRNA["Middle_mean"] + 1)
)

# Sort by strongest Old vs Young change
age_miRNA_sorted = age_miRNA.sort_values(
"log2_Old_vs_Young",
ascending=False
)

print("Age-related analysis completed.")

display(
age_miRNA_sorted[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"log2_Old_vs_Young",
"log2_Old_vs_Middle"
]
].head(20)
)
```

```python
# ==========================================
# STEP 4 — Filter stronger age-associated miRNAs
# ==========================================

# Calculate average expression across all age groups
age_miRNA["Overall_mean"] = age_miRNA[
["Young_mean", "Middle_mean", "Old_mean"]
].mean(axis=1)

# Calculate fold change
age_miRNA["Old_vs_Young_FC"] = (
(age_miRNA["Old_mean"] + 1) /
(age_miRNA["Young_mean"] + 1)
)

# Keep miRNAs with reasonable expression
# and at least 2-fold change in either direction
candidate_miRNAs = age_miRNA[
(age_miRNA["Overall_mean"] >= 10) &
(abs(age_miRNA["log2_Old_vs_Young"]) >= 1)
].copy()

# Sort by magnitude of age-related change
candidate_miRNAs["abs_log2FC"] = abs(
candidate_miRNAs["log2_Old_vs_Young"]
)

candidate_miRNAs = candidate_miRNAs.sort_values(
"abs_log2FC",
ascending=False
)

print("Number of candidate miRNAs:", len(candidate_miRNAs))

display(
candidate_miRNAs[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"log2_Old_vs_Young"
]
].head(30)
)
```

```python
# ==========================================
# STEP 5 — Separate increased and decreased
# ==========================================

# miRNAs increased in Old compared with Young
increased_miRNAs = candidate_miRNAs[
candidate_miRNAs["log2_Old_vs_Young"] >= 1
].sort_values(
"log2_Old_vs_Young",
ascending=False
)

# miRNAs decreased in Old compared with Young
decreased_miRNAs = candidate_miRNAs[
candidate_miRNAs["log2_Old_vs_Young"] <= -1
].sort_values(
"log2_Old_vs_Young",
ascending=True
)

print("Increased in Old:", len(increased_miRNAs))
print("Decreased in Old:", len(decreased_miRNAs))

print("\n===== TOP INCREASED =====")

display(
increased_miRNAs[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"log2_Old_vs_Young"
]
].head(20)
)

print("\n===== TOP DECREASED =====")

display(
decreased_miRNAs[
[
"miRNA_name",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"log2_Old_vs_Young"
]
].head(20)
)
```

```python
# ============================================
# STEP: Build a more reliable candidate list
# ============================================

# Require measurable expression in at least one age group
research_candidates = candidate_df[
(candidate_df["Young_mean"] >= 5) |
(candidate_df["Middle_mean"] >= 5) |
(candidate_df["Old_mean"] >= 5)
].copy()

# Absolute age-associated change
research_candidates["absolute_change"] = (
research_candidates["Old_mean"] -
research_candidates["Young_mean"]
)

# Rank using absolute log2 fold-change
research_candidates["abs_log2_change"] = (
research_candidates["log2_Old_vs_Young"].abs()
)

# Known miRNA vs novel candidate
research_candidates["miRNA_type"] = research_candidates[
"miRNA_name"
].apply(
lambda x: "Novel" if str(x).startswith("novel_") else "Known"
)

# Strong candidates
strong_candidates = research_candidates[
research_candidates["abs_log2_change"] >= 1
].sort_values(
["abs_log2_change", "Old_mean"],
ascending=[False, False]
)

print("Research candidates:", len(strong_candidates))

print("\n===== STRONG AGE-ASSOCIATED miRNAs =====")

display(
strong_candidates[
[
"miRNA_name",
"miRNA_type",
"Young_mean",
"Middle_mean",
"Old_mean",
"Old_vs_Young_FC",
"log2_Old_vs_Young"
]
].head(30)
)
```

```python
# ============================================================
# RESEARCH LAB v3 - STAGE 2: C-ALPHA DISTANCE MATRIX & ANALYSIS
# ============================================================

import numpy as np
from Bio.PDB import PDBParser

print("=" * 70)
print("INITIALIZING RESEARCH LAB v3 - STAGE 2: DISTANCE MATRIX")
print("=" * 70)

# Load the previously downloaded structure file
pdb_id = "1LCD"
structure_file = f"pdb{pdb_id.lower()}.ent" # Standard naming convention from PDBList

parser = PDBParser(QUIET=True)
structure = parser.get_structure(pdb_id, structure_file)

# Extract all C-alpha (CA) atoms from the first model
model = structure[0]
ca_atoms = []
residue_labels = []

for chain in model:
    for residue in chain:
        # Check if C-alpha atom exists in the residue
        if "CA" in residue:
            ca_atoms.append(residue["CA"])
            # Label with Chain ID and Residue Number
            residue_labels.append(f"{chain.id}:{residue.resname}{id(residue) % 1000}")

print(f"Extracted {len(ca_atoms)} C-alpha atoms for distance calculations.")

# Compute pairwise Euclidean distance matrix between C-alpha atoms
num_atoms = len(ca_atoms)
distance_matrix = np.zeros((num_atoms, num_atoms))

for i in range(num_atoms):
    for j in range(num_atoms):
        distance_matrix[i, j] = ca_atoms[i] - ca_atoms[j]

print("-" * 50)
print(f"DISTANCE MATRIX COMPUTED SUCCESSFULLY: shape {distance_matrix.shape}")
print(f"  - Minimum non-zero distance: {np.min(distance_matrix[distance_matrix > 0]):.2f} Å")
print(f"  - Maximum distance: {np.max(distance_matrix):.2f} Å")
print(f"  - Mean pairwise distance: {np.mean(distance_matrix):.2f} Å")
print("-" * 50)

print("=" * 70)
print("RESEARCH LAB v3 STAGE 2 COMPLETED SUCCESSFULLY")
print("=" * 70)
```

```python
# ============================================================
# RESEARCH LAB v3 - STAGE 3: HEATMAP VISUALIZATION
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 70)
print("INITIALIZING RESEARCH LAB v3 - STAGE 3: PLOTTING HEATMAP")
print("=" * 70)

# Set up the matplotlib figure
plt.figure(figsize=(8, 6))

# Plot the distance matrix heatmap using seaborn
sns.heatmap(distance_matrix, cmap="viridis", cbar_kws={'label': 'Inter-atomic Distance (Å)'})

plt.title(f"C-Alpha Distance Matrix Heatmap ({pdb_id})", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Residue Index", fontsize=12)
plt.ylabel("Residue Index", fontsize=12)

plt.tight_layout()
plt.show()

print("-" * 50)
print("Heatmap generated successfully!")
print("-" * 50)
print("=" * 70)
print("RESEARCH LAB v3 FULLY COMPLETED")
print("=" * 70)
```
